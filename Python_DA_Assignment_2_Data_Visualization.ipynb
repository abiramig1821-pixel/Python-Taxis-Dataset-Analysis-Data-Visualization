```python
nonnumerical_col = ['color','payment','pickup_zone','dropoff_zone','pickup_borough','dropoff_borough']
nonnumerical_col
```

```python
# Non numerical columns of missing values are filled using mode() by using for loop and fillna()

for col in nonnumerical_col :
  df[col]=df[col].fillna(df[col].mode()[0])
```

**3) Visualizations using Matplotlib/Pandas Plot:**

  **★ Line Chart**

      Plot a line chart to visualize the fare over time, using the pickup timestamp as the x-axis and fare as the y-axis.
      
      Ensure the pickup column is converted to a datetime format before plotting.

```python
df.dtypes
```

```python
# groupby and avarage fare over the time

daily_trend = df.groupby(df['pickup'].dt.date)['fare'].mean()

plt.figure(figsize=(10,7))
plt.plot(daily_trend.index,daily_trend.values,color='green',marker='o',markerfacecolor='blue',linewidth=2)
plt.xlabel("pickup date")
plt.xticks(rotation=45)
plt.ylabel("Avg fare")
plt.title("Average fare over the pickup time")
plt.tight_layout()
plt.show()
```

**★9 Bar Chart**  

    Create a bar chart to show the total fare for each pickup_borough.
    
    Group the data by pickup_borough and sum the fare for each group.  

```python
# Group by total fare for each pickup_borough

fare_pickup =df.groupby(df['pickup_borough'])['fare'].sum()
fare_pickup
```

```python
plt.bar(fare_pickup.index,fare_pickup.values)
plt.show()
```

```python


plt.figure(figsize=(7,5))
bars=plt.bar(fare_pickup.index,fare_pickup.values)
plt.bar_label(bars)
plt.xlabel("Pickup borough")
plt.ylabel("Total fare")
plt.title("Total fare from pickup borough")
plt.tight_layout()
plt.show()
```

Here, Manhattan pickup borough is higher of total fare followed by Queens, Brooklyn and Bronx.

**★ Pie Chart**

    Plot a pie chart showing the distribution of trips based on the payment method (credit card, cash, etc.).
    
    Each slice should represent the count of trips for a specific payment method.  

```python
# count find
payment_count= df['payment'].value_counts()
payment_count
```

**★ Heatmap**  

    Plot a heatmap to visualize the correlation between numerical variables such as distance, fare, tip, tolls, and total.
    
    Use a correlation matrix to highlight the relationships.  

```python
numerical_col= df.select_dtypes(include=['number']).columns
numerical_col
```

```python
corr_df=df[numerical_col].corr()
corr_df
```

```python
# Heatmap

sns.heatmap(corr_df,annot=True)
plt.show()
```

From the heatmap,
* The relationship between **fare and total** is **0.97** so postive correlation.
* The relationship between **distance, fare and total** is **0.9**.so positive correlation. when distance higher fare and total leads higher.
* The relationshop between **tip, total, and fare** is **0.65 and 0.49**.Its moderate postive correlation.
* The relationshop between **tolls, distance, and total** is **0.64 and 0.68**.Its moderate postive correlation. when distance is long then needs to cross tolls.
* In **passanger** near zero **-0.0029 to 0.021**.So, there is zero correlation.
